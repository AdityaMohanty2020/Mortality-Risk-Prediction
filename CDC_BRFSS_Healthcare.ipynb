 **While the model demonstrates strong baseline performance, minimizing False Negatives is a critical clinical priority. Misclassifying a high-mortality risk patient as low-risk carries life-threatening consequences, whereas the reverse scenario (overestimating risk) is an acceptable trade-off. Consequently, our optimization strategy will strictly prioritize maximizing Recall (Sensitivity) to a target range of 0.99 to 1.00**

```python

y_proba = model.predict_proba(xtest)[:, 1]
y_pred = y_proba>0.3
print(classification_report(ytest, y_pred))
print("ROC-AUC:", roc_auc_score(ytest, y_proba))
print(confusion_matrix(ytest, y_pred))
```
